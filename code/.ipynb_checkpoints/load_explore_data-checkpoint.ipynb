# Import libraries

```python
import numpy as np
import pandas as pd
from pathlib import Path
```

# Load and explore the data

## Load 9 csv files

```python
# Define data directory
data_path = Path("data/raw")

# Load datasets
customers = pd.read_csv(data_path / "olist_customers_dataset.csv")
geolocation = pd.read_csv(data_path / "olist_geolocation_dataset.csv")
order_items = pd.read_csv(data_path / "olist_order_items_dataset.csv")
order_payments = pd.read_csv(data_path / "olist_order_payments_dataset.csv")
order_reviews = pd.read_csv(data_path / "olist_order_reviews_dataset.csv")
orders = pd.read_csv(data_path / "olist_orders_dataset.csv")
products = pd.read_csv(data_path / "olist_products_dataset.csv")
sellers = pd.read_csv(data_path / "olist_sellers_dataset.csv")
category_translation = pd.read_csv(data_path / "product_category_name_translation.csv")
```

## Create dictionary for the dataset

```python
data = {
    "customers": customers,
    "geolocation": geolocation,
    "order_items": order_items,
    "order_payments": order_payments,
    "order_reviews": order_reviews,
    "orders": orders,
    "products": products,
    "sellers": sellers,
    "category_translation": category_translation
}
```

## Check the data dimension

```python
for name, df in data.items():
    print(f"{name}: {df.shape}")
```

## Check data types

#### Summary
1. The geolocation dataset contains the largest number of records (1,000,163), followed by order items (112,650).

## Check missing values

```python
for name, df in data.items():
    print(f"\n{name}")
    print(df.isnull().sum())
```

#### Summary
1. Most datasets have no missing values, while missing data is mainly concentrated in the order reviews, orders, and products datasets.
2. The orders dataset contains missing delivery-related timestamps, including 2,965 missing values in order_delivered_customer_date. These records require further investigation before analyzing delivery performance.
3. The products dataset has 610 missing values in several product description fields and only two missing values in each weight and dimension field.

## Check duplicates

```python
for name, df in data.items():
    print(f"{name}: {df.duplicated().sum()} duplicates")
```

#### Summary
1. Eight of the nine datasets contain no exact duplicate rows.
2. The geolocation dataset contains 261,831 duplicate rows, representing approximately 26.2% of its records.

## Explore Categorical Variables

### Order status

```python
orders["order_status"].value_counts()
```

Most orders have been successfully delivered. The remaining orders are distributed across other statuses, including shipped, canceled, unavailable, and processing. Order status should be considered when analyzing delivery performance because not all orders have completed the delivery process.

### Product category name

```python
products["product_category_name"].value_counts().head(10)
```

The three most common product categories are bed, bath & table (3,029 products), sports & leisure (2,867), and furniture & decoration (2,657).
We might need to translate product_category_name into English in the data processing step.

### Customer state

```python
customers["customer_state"].value_counts().head(10)
```

São Paulo (SP) has the highest number of customer records (41,746), accounting for approximately 42% of the customers dataset.
Rio de Janeiro (RJ) and Minas Gerais (MG) follow with 12,852 and 11,635 records, respectively. Customer records are geographically concentrated in a few states, which may be relevant when analyzing regional purchasing patterns and delivery performance.

### Payment type

```python
order_payments["payment_type"].value_counts()
```

Credit cards are the most frequently recorded payment method, accounting for approximately 73.9% of payment records. Boleto is the second most common payment method, followed by vouchers and debit cards.

### Order reviews

```python
order_reviews["review_score"].value_counts().sort_index()
```

Customer reviews are generally positive, with approximately 77.1% receiving scores of 4 or 5. Five-star reviews are the most common, accounting for 57,328 of the 99,224 review records. Low ratings (1–2 stars) account for approximately 14.7% of review records.

The distribution is imbalanced toward positive reviews, which should be considered if review scores are used for classification modeling.

## Explore Relationships Between Tables

### Check unique IDs

```python
# Customers
print("Unique customer IDs:", customers["customer_id"].nunique())
print("Unique customers:", customers["customer_unique_id"].nunique())

# Orders
print("Unique orders:", orders["order_id"].nunique())

# Order Items
print("Unique orders in order_items:", order_items["order_id"].nunique())
print("Unique products in order_items:", order_items["product_id"].nunique())

# Products
print("Unique products:", products["product_id"].nunique())

# Sellers
print("Unique sellers:", sellers["seller_id"].nunique())

# Payments
print("Unique orders in payments:", order_payments["order_id"].nunique())

# Reviews
print("Unique orders in reviews:", order_reviews["order_id"].nunique())
```

#### Summary
1. The customers dataset contains 99,441 customer IDs but only 96,096 unique customer identifiers, indicating that some customers are associated with multiple customer records.
2. The orders dataset contains 99,441 unique orders, but only 98,666 appear in the order items dataset.
3. The payment and review datasets contain records for 99,440 and 98,673 unique orders, respectively.

### Check number of items per order

```python
items_per_order = order_items.groupby("order_id").size()
items_per_order.describe()
```

#### Summary
1. Most orders contain only one item, with a median of 1 item per order.
2. The average is approximately 1.14 items per order, while the maximum is 21.
3. This confirms that the order items dataset has a one-to-many relationship with the orders dataset.

### Check number of payments per order

```python
payments_per_order = order_payments.groupby("order_id").size()
payments_per_order.describe()
```

#### Summary
1. Most orders have a single payment record, with a median of 1.
2. Some orders have multiple payment records, with a maximum of 29.
3. Payment records may need to be aggregated at the order level before merging to prevent duplicated order information.

### Check number of reviews per order

```python
reviews_per_order = order_reviews.groupby("order_id").size()
reviews_per_order.describe()
```

#### Summary
1. Most reviewed orders have only one review record.
2. A small number of orders have multiple review records, with a maximum of three.
3. If customer satisfaction is analyzed at the order level, multiple reviews for the same order will need to be handled consistently.

# Conclusion
1. The datasets contain information about 99,441 orders, 96,096 unique customers, 32,951 products, and 3,095 sellers. Common identifiers allow the datasets to be connected for further analysis.
2. Most datasets have no missing values or exact duplicate rows. However, missing values are present in order reviews, delivery timestamps, and product attributes. The geolocation dataset also contains 261,831 duplicate rows that require further investigation.
3. Approximately 97% of orders have been delivered. Customer records are concentrated in São Paulo, Rio de Janeiro, and Minas Gerais, while credit cards are the most frequently recorded payment method.
4. Customer reviews are generally positive, with approximately 77% receiving ratings of 4 or 5. This distribution may present a class imbalance challenge for future classification models.
5. Most orders contain a single item, payment record, and review. However, some orders have multiple associated records, which must be considered during data integration.
